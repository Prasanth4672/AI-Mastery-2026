```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.ndimage import convolve
```

```python
# Load a sample gray-scale image (for example, a simple 5x5 image)
image = np.random.rand(10, 10)
print("Original Image:\n", image)
```

```python
# Define convolution filters (kernels)
edge_detection_kernel = np.array([[ -1, -1, -1],
                                  [ -1,  8, -1],
                                  [ -1, -1, -1]])   
blur_kernel = np.array([[ 1/9, 1/9, 1/9],
                        [ 1/9, 1/9, 1/9],
                        [ 1/9, 1/9, 1/9]])
sharpen_kernel = np.array([[ 0, -1,  0],
                           [-1,  5, -1],
                           [ 0, -1,  0]])   
# Apply convolution with the defined filters
edge_detected_image = convolve(image, edge_detection_kernel)
blurred_image = convolve(image, blur_kernel)
sharpened_image = convolve(image, sharpen_kernel)
# Display the original and processed images
plt.figure(figsize=(12, 4))
plt.subplot(1, 4, 1)
plt.title("Original Image")
plt.imshow(image, cmap='gray')
plt.subplot(1, 4, 2)
plt.title("Edge Detected Image")
plt.imshow(edge_detected_image, cmap='gray')
plt.subplot(1, 4, 3)
plt.title("Blurred Image")
plt.imshow(blurred_image, cmap='gray')
plt.subplot(1, 4, 4)
plt.title("Sharpened Image")
plt.imshow(sharpened_image, cmap='gray')
plt.show()
```

```python
import tensorflow as tf
```

```python
# Create a sample input tensor(batch_size, height, width, channels)
input_tensor = tf.random.normal([1, 10, 10, 1])

# Define a simple convolutional layer
conv_layer = tf.keras.layers.Conv2D(filters=1, kernel_size=(3, 3), padding='same', strides=(1, 1))

# Apply the convolutional layer to the input tensor
output_tensor = conv_layer(input_tensor)

print("Input Tensor Shape:", input_tensor.shape)
print("Output Tensor Shape:", output_tensor.shape)
```

```python
import torch
import torch.nn as nn
```

```python
# Create a sample input tensor(batch_size, channels, height, width)
input_tensor_pt = torch.randn(1, 1, 10, 10)  # PyTorch uses (batch_size, channels, height, width)

#Define a simple convolutional layer
conv_layer_pt = nn.Conv2d(in_channels=1, out_channels=1, kernel_size=3, padding=1, stride=1)

# Apply the convolutional layer to the input tensor
output_tensor_pt = conv_layer_pt(input_tensor_pt)

print("Input Tensor Shape:", input_tensor_pt.shape)
print("Output Tensor Shape:", output_tensor_pt.shape)
```
