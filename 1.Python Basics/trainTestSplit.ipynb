```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.model_selection import train_test_split

```

```python
## Load the dataset tips.csv
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
tips = pd.read_csv(url)
#Inspect Data
print(tips.info())
print(tips.describe())
```

```python
# Define features and target variable
features = tips[['total_bill', 'size']]
target = tips['tip']

print(features.head())
print(target.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)
print(X_train.shape, X_test.shape)
print(y_train.shape, y_test.shape)
```

```python
# Visualize the Relationships between Features and Target
sns.pairplot(tips, x_vars=['total_bill', 'size'], y_vars='tip', height=5, aspect=0.8, kind='scatter')
plt.suptitle('Scatter Plots of Total Bill and Size vs Tip', y=1.02)
plt.show()
```
