```python
from itertools import product

# Sample space of dice rolls
sample_space = list(range(1, 7))

#Probability of even number on the die
even_numbers = [2, 4, 6]
probability_even = len(even_numbers) / len(sample_space)
print(f"Probability of rolling an even number: {probability_even:.2f}")
```

```python
import numpy as np

#Random variable:  dice roll
outcomes = np.array([1, 2, 3, 4, 5, 6])
probabilites = np.array([1/6] * 6)  # Uniform distribution

#Expectation 
expectation = np.sum(outcomes * probabilites)
print(f"Expected value of a dice roll: {expectation}")

#Variance and Standard Deviation
variance = np.sum(probabilites * (outcomes - expectation) ** 2)
std_dev = np.sqrt(variance)
print(f"Variance of a dice roll: {variance:.2f}")
print(f"Standard Deviation of a dice roll: {std_dev:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Setup parameters
mu = 0       # Mean (center)
sigma = 0.1  # Standard deviation (spread)

# 2. Generate 1000 random samples from a Normal Distribution
samples = np.random.normal(mu, sigma, 1000)

# 3. Create the plot
count, bins, ignored = plt.hist(samples, 30, density=True, color='skyblue', edgecolor='black')

# 4. Plot the theoretical probability density function (the curve)
plt.plot(bins, 1/(sigma * np.sqrt(2 * np.pi)) *
         np.exp( - (bins - mu)**2 / (2 * sigma**2) ),
         linewidth=2, color='r')

plt.title("Normal Distribution (Mean=0, StdDev=0.1)")
plt.show()
```
