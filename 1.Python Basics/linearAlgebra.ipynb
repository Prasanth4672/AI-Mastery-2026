```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Matrix addition
C = A + B
print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("A + B:\n", C)

# Matrix Subtraction
D = A - B
print("A - B:\n", D)

# Matrix Multiplication
E = np.dot(A, B)
print("A * B:\n", E)
```

```python
# Identity Matrix
I = np.eye(4)
print("Identity Matrix:\n", I)
```

```python
#Zero Matrix
Z = np.zeros((3, 3))
print("Zero Matrix:\n", Z)
```

```python
#Diagonal Matrix
D = np.diag([1, 2, 3])
print("Diagonal Matrix:\n", D)
```

```python
# Matrix - Vector Multiplication
v = np.array([1, 2])
result = np.dot(A, v)
print("Matrix A:\n", A)
print("Vector v:\n", v)
print("A * v:\n", result)   
```
